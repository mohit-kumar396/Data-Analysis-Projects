```python
# Q1. Prfit/loss on each product

df["profit_loss"] = df["SP"] - df["CP"]
df
# Positive Value=profit
# Negative Value = Loss
# 0 = No profit no loss
```

```python
# Q2. Prfit_percent/loss_percent on each product

df["profit_loss_percent"] = (df["profit_loss"]/df["CP"]) * 100
df
```

```python
# Q3. product corresponding to the maximum Profit_Loss value.

df.loc[df["profit_loss"].idxmax(), "Product"]

# Syntax Of the above code
# DataFrame.loc[DataFrame["column"].idxmax(), "target_column"]
# df["Profit_Loss"].idxmax() :- Profit_Loss column mein maximum value ka index label return karta hai.
# df.loc[index, "Product"] :- us index ki "Product" value return karta hai.
```

```python
# Q10. Net profit/loss

Net_profit_loss= df["profit_loss"].sum()
Net_profit_loss
```

```python
# Q11. Lets increase sp by 10% and see what will be the net_profit?

df["new_sp"]=df["SP"] * 1.1
df["new_profit_loss"]=df["new_sp"]-df["CP"]
new_net_profit_loss= df["new_profit_loss"].sum()
new_net_profit_loss
```

```python
# Q12. Overall profit increase after increasing sp by 10% is
new_profit_increase=new_net_profit_loss-Net_profit_loss
new_profit_increase
```

```python
# Final Insight:-

# "Laptop has the highest absolute profit, but Backpack, USB Hub, Desk Lamp, T-Shirt has a higher profit margin.
# This suggests that Laptop generates more profit per unit,
# while Backpack, USB Hub, Desk Lamp, T-Shirt provides better profitability relative to its cost."
```

```python
# High profit + high sales → ⭐ Strong product
# High profit % + high sales → ⭐ Strong and efficient product
# High profit but low sales → Try increasing sales
# High profit % but low sales → Try increasing demand/sales
# Low profit + low sales → Consider improving or dropping the product
```

```python
import pandas as pd
import sqlite3
profit_loss_df=pd.read_csv("profit.csv")
conn=sqlite3.connect(":memory:")
```

```python
profit_loss_df.to_sql("profit", conn,index=False, if_exists="replace")
df
```

```python
# Q1. Display all products with their Category, Cost Price, Selling Price
```

```python
query1="""
select p.category, p.CP, p.SP
from profit p
"""
result=pd.read_sql_query(query1, conn)
result
```

```python
# Q2. Find all products that belong to the Electronics category.
```

```python
query2="""
select p.Product as Electronic_product
from profit p
where p.Category="Electronics"
"""
result=pd.read_sql_query(query2, conn)
result
```

```python
# Q3. Find all products where the Selling Price is greater than the Cost Price.
```

```python
query3="""
select p.Product as Product_with_SP_greater_than_CP
from profit p
where p.SP>p.CP
"""
result=pd.read_sql_query(query3, conn)
result
```

```python
#Q4. Find all products where the Selling Price is less than the Cost Price.
```

```python
query4="""
select p.Product as Product_with_SP_less_than_CP
from profit p
where p.SP<p.CP
"""
result=pd.read_sql_query(query4, conn)
result
```

```python
# Q5.Display all products in descending order of Selling Price.
```

```python
query5="""
select p.Product as Product_in_descending_order_of_SP
from profit p
order by p.SP desc
"""
result=pd.read_sql_query(query5, conn)
result
```
